```python
df = pd.DataFrame(student_data, columns=columns)

print("Shape (rows, columns):", df.shape)
print("\nColumns:", list(df.columns))
print("\nFirst 5 records:")
display(df.head())

print("\nBasic information:")
df.info()

print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
PASS_MARK = 40   # minimum marks needed in EVERY subject to pass

def calculate_total(marks):
    """Total marks per student (NumPy, row-wise)."""
    return np.sum(marks, axis=1)

def calculate_average(marks):
    """Average marks per student, rounded to 2 decimals."""
    return np.round(np.mean(marks, axis=1), 2)

def assign_grade(average, lowest_mark):
    """
    Grading rule:
      Any subject below 40 -> F
      90+ -> A+, 80-89 -> A, 70-79 -> B, 60-69 -> C, 50-59 -> D, 40-49 -> E
    """
    if lowest_mark < PASS_MARK:
        return "F"
    elif average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "E"

def check_result(subject_marks):
    """Pass only if the student scores at least PASS_MARK in every subject."""
    return "Pass" if np.all(subject_marks >= PASS_MARK) else "Fail"

print("Functions defined successfully.")
print("Grading rule: A+ (90+), A (80+), B (70+), C (60+), D (50+), E (40+), F (any subject < 40)")
print("Pass condition: at least", PASS_MARK, "marks in every subject")
```

```python
marks_array = df[subjects].to_numpy()

df["Total"] = calculate_total(marks_array)
df["Average"] = calculate_average(marks_array)

grades = []
results = []
for i in range(len(df)):
    student_marks = marks_array[i]
    grades.append(assign_grade(df.loc[i, "Average"], student_marks.min()))
    results.append(check_result(student_marks))

df["Grade"] = grades
df["Result"] = results

df["Rank"] = df["Average"].rank(ascending=False, method="min").astype(int)

display(df)
```

```python
total_students = len(df)
passed = (df["Result"] == "Pass").sum()
failed = (df["Result"] == "Fail").sum()

top_student = df.loc[df["Average"].idxmax()]
low_student = df.loc[df["Average"].idxmin()]

print("Total students       :", total_students)
print("Students passed      :", passed)
print("Students failed      :", failed)
print("Pass percentage      :", round(passed / total_students * 100, 2), "%")
print("Overall class average:", round(df["Average"].mean(), 2))
print("Highest average      :", top_student["Name"], "-", top_student["Average"])
print("Lowest average       :", low_student["Name"], "-", low_student["Average"])
print("Average attendance   :", round(df["Attendance"].mean(), 2), "%")

print("\nGrade distribution:")
print(df["Grade"].value_counts().sort_index())

print("\nFailed students:")
display(df[df["Result"] == "Fail"][["Student ID", "Name", "Department"] + subjects + ["Average"]])
```

```python
subject_report = []

for subject in subjects:
    col = df[subject].to_numpy()
    subject_report.append({
        "Subject": subject,
        "Highest": np.max(col),
        "Lowest": np.min(col),
        "Average": round(np.mean(col), 2),
        "Students Passed": int(np.sum(col >= PASS_MARK)),
        "Students Failed": int(np.sum(col < PASS_MARK)),
    })

subject_df = pd.DataFrame(subject_report)
display(subject_df)

best = subject_df.loc[subject_df["Average"].idxmax()]
weak = subject_df.loc[subject_df["Average"].idxmin()]
print("Best-performing subject :", best["Subject"], "(average", best["Average"], ")")
print("Weakest subject         :", weak["Subject"], "(average", weak["Average"], ")")
```

```python
top_5 = df.sort_values("Average", ascending=False).head(5)

print("Top 5 performing students:\n")
display(top_5[["Rank", "Student ID", "Name", "Department", "Total", "Average", "Grade"]])

print("\nTopper in each subject:")
for subject in subjects:
    row = df.loc[df[subject].idxmax()]
    print(f"  {subject:<18}: {row['Name']} ({row[subject]})")
```

```python
dept_report = df.groupby("Department").agg(
    Students=("Name", "count"),
    Avg_Marks=("Average", "mean"),
    Avg_Attendance=("Attendance", "mean"),
    Passed=("Result", lambda x: (x == "Pass").sum()),
).round(2)

display(dept_report)
print("Best department by average marks:", dept_report["Avg_Marks"].idxmax())
```
